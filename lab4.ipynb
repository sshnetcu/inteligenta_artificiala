```python
df.columns = df.columns.str.strip()

df['Value'] = df['Value'].replace('[€,]', '', regex=True)
df['Wage'] = df['Wage'].replace('[€,]', '', regex=True)

df['Value'] = df['Value'].replace({'K': '*1e3', 'M': '*1e6'}, regex=True).map(pd.eval)
df['Wage'] = df['Wage'].replace({'K': '*1e3', 'M': '*1e6'}, regex=True).map(pd.eval)
```

```python
df['is_underpaid'] = df['Wage'] < (df['Value'] / 100)
```

```python
df['Score'] = (
    0.3 * df['Overall'] +
    0.3 * df['Potential'] +
    0.2 * df['SprintSpeed'] +
    0.2 * df['ShortPassing']
)
```

```python
good_deals = df[['Name', 'Wage', 'Value']].copy()
good_deals['difference'] = good_deals['Value'] - good_deals['Wage']

good_deals.sort_values(by='difference', ascending=False)
```
