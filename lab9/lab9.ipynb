```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Flatten, Dense
import matplotlib.pyplot as plt
import numpy as np
```

```python
mnist = tf.keras.datasets.mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalizare
x_train = x_train / 255.0
x_test = x_test / 255.0

print("Train:", x_train.shape)
print("Test:", x_test.shape)
```

```python
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_data=(x_test, y_test)
)
```

```python
test_loss, test_acc = model.evaluate(x_test, y_test)

print("Acuratete:", test_acc)
```

```python
predictions = model.predict(x_test)

index = 0

plt.imshow(x_test[index], cmap='gray')
plt.title(f"Eticheta reala: {y_test[index]}")
plt.axis('off')
plt.show()

predicted_label = np.argmax(predictions[index])

print("Predictia modelului:", predicted_label)
```

```python
fashion_mnist = tf.keras.datasets.fashion_mnist

(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
fashion_model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

fashion_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

fashion_model.fit(x_train, y_train, epochs=5)

loss_fashion, acc_fashion = fashion_model.evaluate(x_test, y_test)

print("Acuratete Fashion MNIST:", acc_fashion)
```

```python
class_names = [
    'T-shirt/top',
    'Trouser',
    'Pullover',
    'Dress',
    'Coat',
    'Sandal',
    'Shirt',
    'Sneaker',
    'Bag',
    'Ankle boot'
]

predictions = fashion_model.predict(x_test)

index = 5

plt.imshow(x_test[index], cmap='gray')
plt.title(f"Eticheta reala: {class_names[y_test[index]]}")
plt.axis('off')
plt.show()

predicted_label = np.argmax(predictions[index])

print("Predictia modelului:", class_names[predicted_label])
```
