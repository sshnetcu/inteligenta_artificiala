```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
```

```python
mnist = keras.datasets.mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
print("Shape train:", x_train.shape)
print("Shape test:", x_test.shape)

print("Prima eticheta:", y_train[0])
```

```python
plt.imshow(x_train[0], cmap="gray")
plt.title(f"Eticheta: {y_train[0]}")
plt.colorbar()
plt.show()
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),

    keras.layers.Dense(128, activation='relu'),

    keras.layers.Dense(10, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

Antrenament

```python
history = model.fit(
    x_train,
    y_train,
    epochs=5
)
```

Evaluare

```python
test_loss, test_acc = model.evaluate(x_test, y_test)

print("Acuratete pe test:", test_acc)
```

```python
predictions = model.predict(x_test)
```

```python
index = 0

plt.imshow(x_test[index], cmap="gray")
plt.title("Imagine test")
plt.show()

predicted_label = np.argmax(predictions[index])

print("Predictie model:", predicted_label)
print("Eticheta reala:", y_test[index])
```

## Fashion MNIST

```python
fashion_mnist = keras.datasets.fashion_mnist

(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
class_names = [
    'T-shirt/top',
    'Trouser',
    'Pullover',
    'Dress',
    'Coat',
    'Sandal',
    'Shirt',
    'Sneaker',
    'Bag',
    'Ankle boot'
]
```

```python
fashion_model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

fashion_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
fashion_model.fit(
    x_train,
    y_train,
    epochs=5
)
```

```python
test_loss, test_acc = fashion_model.evaluate(x_test, y_test)

print("Acuratete Fashion MNIST:", test_acc)
```

```python
predictions = fashion_model.predict(x_test)

index = 0

plt.imshow(x_test[index], cmap='gray')
plt.title("Imagine Fashion")
plt.show()

predicted_class = np.argmax(predictions[index])

print("Predictie:", class_names[predicted_class])
print("Eticheta reala:", class_names[y_test[index]])
```
